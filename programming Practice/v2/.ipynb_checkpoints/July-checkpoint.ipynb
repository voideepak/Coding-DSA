## Binary Search

```python
def binarySearch(arr, target):
    l = 0
    r = len(arr) - 1
    while l < r:
        mid = (l+r)//2
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            r -= 1
        else:
            l+=1
    return -1

arr = [1,2,3,4,5,6,7]
target = 6
binarySearch(arr,target)
```
